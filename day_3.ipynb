```python
#list
music=["shape of you","taki taki","on my way"]
food=["biriyani","chiken 65","chiken gravy"]
juice=["pineapple","watermelone","muskmelone"]

print("my fav music", music)
print("my fav food", food)
print("my fav juice", juice)
```

```python
#append
music.append("shape of you")
print(music)
```

```python
#insert
music.insert(1,"perfect")
print("after insertion", music)
```

```python
#remove
music.remove("perfect")
print("after remove", music)
```

```python
#pop
music.pop()
print("after pop", music)
```

```python
#reverse
music.reverse()
print("after reverse", music)
```

```python
#list slicing
print("top 2 songs",music[0:2])
print("last 2 songs",music[-2:])
```

```python
#iteration
for i in music:
    print(i)
```

```python
if "shape of you" in music:
    print("yes")
else:
    print("no")
```
